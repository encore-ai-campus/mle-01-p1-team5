```python
import pandas as pd

# Retriever 결과 읽기
df = pd.read_csv(
    "retriever_results.csv"
)

# 질문별 한 행만 남기기
gold_template = (
    df[
        [
            "query_id",
            "question",
        ]
    ]
    .drop_duplicates(subset=["query_id"])
    .sort_values("query_id")
    .reset_index(drop=True)
)

# 사람이 작성할 컬럼 추가
gold_template["gold_chunks"] = ""
gold_template["evidence"] = ""

# 저장
gold_template.to_csv(
    "data/gold_set.csv",
    index=False,
    encoding="utf-8-sig",
)

print(gold_template)
```
